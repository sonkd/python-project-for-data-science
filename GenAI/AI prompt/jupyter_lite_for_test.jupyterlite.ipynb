```python
from pyodide.http import pyfetch

async def download(url, filename):
    response = await pyfetch(url)
    if response.status == 200:
        with open(filename, "wb") as f:
            f.write(await response.bytes())

path = URL

await download(path, "dataset.csv")
file_name  = "dataset.csv"
```

```python
#This script reads a CSV into a Pandas DataFrame, computes which columns have any missing values using isna().any(), and prints the list of such columns along with a per-column missing-value count.

import pandas as pd

# Path to the CSV file
file_path = "dataset.csv"

# Read the CSV into a DataFrame (first row treated as headers by default)
df = pd.read_csv(file_path)

# Identify columns that contain any missing values
cols_with_missing = df.columns[df.isna().any()].tolist()

# Output the list of columns with missing values
print("Columns with missing values:", cols_with_missing)

# Optional: display missing value counts per column
missing_counts = df.isna().sum()
print("Missing values per column:")
print(missing_counts)

# Assume df is your existing DataFrame
df['Screen_Size_cm'] = df['Screen_Size_cm'].fillna(df['Screen_Size_cm'].mode().iloc[0])
df['Weight_kg'] = df['Weight_kg'].fillna(df['Weight_kg'].mean())

# Convert the specified columns to float
df['Screen_Size_cm'] = df['Screen_Size_cm'].astype(float)
df['Weight_kg'] = df['Weight_kg'].astype(float)

# Convert cm to inches and rename
df['Screen_Size_inch'] = df['Screen_Size_cm'] / 2.54
df = df.drop(columns=['Screen_Size_cm'])

# Convert kg to pounds and rename
df['Weight_pounds'] = df['Weight_kg'] * 2.2046226218
df = df.drop(columns=['Weight_kg'])

max_val = df['CPU_frequency'].max()
df['CPU_frequency'] = df['CPU_frequency'] / max_val

# 1. Create indicator variables for Screen into df1
df1 = pd.get_dummies(df['Screen'], prefix='Screen')

# 2. Append df1 into the original DataFrame
df = pd.concat([df, df1], axis=1)
# df = df.join(df1)

# 3. Drop the original Screen attribute
df.drop(columns=['Screen'], inplace=True)
# df.drop(columns=['Screen'], inplace=True)

usd_to_eur = 0.92
df['Price'] = df['Price'] * usd_to_eur

# Min-max normalize CPU_frequency in place
min_val = df['CPU_frequency'].min()
max_val = df['CPU_frequency'].max()
df['CPU_frequency'] = (df['CPU_frequency'] - min_val) / (max_val - min_val)
```
